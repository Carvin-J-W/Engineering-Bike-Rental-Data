# Citibike Data Cleaning/Inspection

```python
jan_data = pd.read_csv("JC-201601-citibike-tripdata.csv")
feb_data = pd.read_csv("JC-201602-citibike-tripdata.csv")
mar_data = pd.read_csv("JC-201603-citibike-tripdata.csv")
apr_data = pd.read_csv("JC-201604-citibike-tripdata.csv")
may_data = pd.read_csv("JC-201605-citibike-tripdata.csv")
jun_data = pd.read_csv("JC-201606-citibike-tripdata.csv")
jul_data = pd.read_csv("JC-201607-citibike-tripdata.csv")
aug_data = pd.read_csv("JC-201608-citibike-tripdata.csv")
sep_data = pd.read_csv("JC-201609-citibike-tripdata.csv")
oct_data = pd.read_csv("JC-201610-citibike-tripdata.csv")
nov_data = pd.read_csv("JC-201611-citibike-tripdata.csv")
dec_data = pd.read_csv("JC-201612-citibike-tripdata.csv")

full_year_list = [jan_data, feb_data, mar_data, apr_data, may_data, jun_data, jul_data, aug_data, sep_data, oct_data, nov_data, dec_data]

combined_df = pd.concat(full_year_list, ignore_index=True) 
```
